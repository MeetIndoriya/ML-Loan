## Data Preprocessing

```python
import matplotlib as plt
```

```python
df.isnull().sum()
# print("\n\n\nPercentage of Outlier:",df.isnull().mean() * 100)
```

```python
df['LoanAmount'].fillna(df['LoanAmount'].median(), inplace=True)
```

```python
import matplotlib.pyplot as plt

education = df['Education'].value_counts()
plt.bar(education.index, education.values, color='skyblue', edgecolor='black')
plt.title('Borrower Count by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Number of Borrowers')
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.hist(df['Income'], bins=80, color='yellow', edgecolor='black')
plt.title('Histogram of Income')
plt.xlabel('Income')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.boxplot(df['Income'])
plt.title('Box Plot of Income')
plt.ylabel('Income')
```
